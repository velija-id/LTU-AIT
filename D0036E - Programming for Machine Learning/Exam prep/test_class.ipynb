```python
from charset_normalizer import cd


class Hypercube:
    def __init__(self, d, l):
        self.dimension = d
        self.length = l

    def whatAmI(self):
        print("I am a ", self.dimension, " dimensional hypercube")

class Square(Hypercube):
    def __init__(self, l):
        super().__init__(d=2, l=l)
        self.edges = 4
    def area(self):
        return self.length * self.length
    def perimeter(self):
        return self.length*self.edges
```

```python
my_cube = Hypercube(d=3, l=4)
my_cube.whatAmI()
```

```python
my_square = Square(l=7)
my_square.whatAmI()
print(my_square.area())
print(my_square.perimeter())
```

```python
def isSorted(list):
    sorted = True
    for idx, i in enumerate(list):
        if idx == len(list) -1:
            break
        if list[idx] > list[idx+1]:
            sorted = False
    return sorted
```

```python
my_list = [0,3,1,2]
isSorted(my_list)
```

```python
def countNegativeCorrected(lst):
    i=0
    j=0
    while i<=len(lst):
        if lst[i]<0:
            i=i+1
            j=j+1
        else:
            i=i+1
            if i == len(lst):
                break
    return j
```

```python
def insert(lst, e):
    i=0
    while lst[i]<e:
        i = i+1
    lst[i:i] = [e]
def splitIfFull(lst):
    if len(lst)>4:
        return (lst[:3],lst[3:])
    else:
        return (lst,[])
```

```python
def findMul(s, c):
    if c not in s:
        return s
    occurence = 1
    new_word = ""
    for i in range(len(s)):
        if s[i] == c:
            new_word = new_word + c*occurence
            occurence = occurence + 1
        else:
            new_word = new_word + s[i]
    return new_word
```

```python
def increaseCounter(filename: str):
    try:
        f=open(filename,'r')
    except FileNotFoundError as e:
        raise ValueError("File not found") from e
    try:
        v=int(f.readline())
    except ValueError as e:
        raise ValueError("Increase counter error") from e
    f.close()
    f=open(filename,'w')
    f.write(str(v+1)+'\n')
    f.close()
def fileCounter():
    file_name = input("Enter file name: ")
    while input("Press enter to count.")=="":
        try:
            increaseCounter(file_name)
        except ValueError:
            print ("counter error")
```
